```python
import numpy as np
import random as rnd
import matplotlib.pyplot as plt

# --- --- --- --- ---
#Funciones:

#Vértice aleatorio entre los tres proporcionados
def vértice(v1, v2, v3):
  v = rnd.choice((v1, v2, v3))
  return v

#rnd.choice(cosa)

#Punto medio entre dos puntos
def pto_medio(p1, p2):
  pm = ((p2[0] + p1[0])/2), ((p2[1] + p1[1])/2)
  return pm
```

```python
# --- --- --- --- ---
#Diseño del triángulo principal:

#Vértices del triángulo
vértice_1 = np.array([0, 0])
vértice_2 = np.array([6, 0])
vértice_3 = np.array([3, 3*np.sqrt(3)])

#Rango de área del triángulo dentro de los ejes
rango_x = np.linspace(0, 6, 100)
rango_y = np.linspace(0, 3*np.sqrt(3), 100)

#Malla formada por el rango
X, Y = np.meshgrid(rango_x, rango_y)

#Funciones que delimitan el triángulo
recta1 = np.sqrt(3)*X
recta2 = -(np.sqrt(3)*X) + 6*np.sqrt(3)

#Creando una máscara para formar el triángulo interno en la malla
máscara = (Y<recta1) & (Y<recta2) & (Y>0)

#Aplicando la máscara a los puntos
X_tri_int = X[máscara]
Y_tri_int = Y[máscara]

#plt.plot(X_tri_int, Y_tri_int, ".", color = "red")
#plt.title("Malla triangular interna")
#plt.show()

#Juntando las parejas
pts_tri_int = np.column_stack((X_tri_int, Y_tri_int))

#print(pts_tri_int)
```

```python
# --- --- --- --- ---
#Generando el triángulo de Siperpiński:

#Seleccionando al azar un punto dentro del triángulo
pto_0 = rnd.choice(pts_tri_int)

#Generando lista de puntos para formar el triángulo de Siperpiński
pts = [pto_0]

while len(pts)<10001:
  pto_i = pto_medio(pts[-1], vértice(vértice_1, vértice_2, vértice_3))
  pts.append(pto_i)

pts_np = np.array(pts)
#print(pts_np)

#Separando los puntos por su eje de coordenadas
x_obt = pts_np[:, 0]
y_obt = pts_np[:, 1]

# --- --- --- --- ---
#Graficando
plt.figure(figsize = (6, 5))

plt.plot(x_obt, y_obt, '.', markersize=1)
plt.title('Triángulo de Siperpiński')

plt.xlabel('X')
plt.ylabel('Y')
#plt.grid(True)

plt.show()
```
