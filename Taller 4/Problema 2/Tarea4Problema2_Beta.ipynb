&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; Generador de números aleatorios usando el método lineal congruente:

- Utilizando los parámetros:
$$
(r_1, a, c, M) = (10, 57, 1, 256)
$$

```python
import numpy as np
import random as rnd
import matplotlib.pyplot as plt

#Creación de la función
def lineal_congruente(r, a, c, M, n): #semilla, multiplicador, incremento, módulo, cantidad límite de rgn
  r_0 = r
  r_n = [r_0]

  while True:
    r_nuevo = ((a*r_n[-1]) + c)%M

    if r_nuevo == r_0 or len(r_n)>(n-1):
      break

    r_n.append(r_nuevo)
  return r_n

#Registro de resultados para los parámetros solicitados
resultados_A = lineal_congruente(10, 57, 1, 256, 256)
#print(resultados)

#¿Cuántos números se generan antes de que la secuencia se repita?
print(f"Antes de que la secuencia se repita, se generan {len(resultados_A)} números.")
```

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; Analizando correlaciones de pares sucesivos $(x_i, y_i) = (r_{2i−1},r_{2i})$:

```python
#Creación de arreglos x_i_A, y_i_A
x_i_A = []
y_i_A = []

for i in range(len(resultados_A)):
  residuo = i%2

  if residuo == 0:
    y_i_A.append(resultados_A[i])
  elif residuo != 0:
    x_i_A.append(resultados_A[i])

#print(x_i_A)
#print(y_i_A)


#Gráfico de (x_i_A, y_i_A)
plt.plot(x_i_A, y_i_A, ".")
plt.title("Correlación de pares sucesivos: método lineal congruente (A)")

plt.xlabel("x")
plt.ylabel("y")

plt.grid()
plt.show()
```

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; Probando el generador de números aleatorios del lenguaje:

```python
#Registro de resultados utilizando 'random'
resultados_random = []

for i in range(0, 256):
  rgn = rnd.choice(range(0, 256))
  resultados_random.append(rgn)

#print(resultados_random)


#Creación de arreglos x_r, y_r
x_r = []
y_r = []

for i in range(len(resultados_random)):
  residuo = i%2

  if residuo == 0:
    y_r.append(resultados_random[i])
  elif residuo != 0:
    x_r.append(resultados_random[i])

#print(x_r)
#print(y_r)


#Gráfico de (x_r, y_r)
plt.plot(x_r, y_r, ".")
plt.title("Correlación de pares sucesivos: utilizando 'random'")

plt.xlabel("x")
plt.ylabel("y")

plt.grid()
plt.show()
```

- Utilizando los parámetros:
$$
(r_1, a, c, M) = (10, 5DEECE66D_{\text{(base 16)}}, 11, 2^{48})
